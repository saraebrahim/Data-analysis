<h2>Data Cleaning</h2>

```python
# Find all the null values in the data set 
df.isnull()
```

```python
# add the numebr of null values in each row
df.isnull().sum()
```

```python
# fill the cylinder column's null values with its mean value
df["Cylinders"] = df["Cylinders"].fillna(df["Cylinders"].mean())
df
```

```python
df.isnull().sum()
```
